```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten
from keras.datasets import mnist
```

```python
(X_train, y_train),(X_test, y_test) = mnist.load_data()
```

```python
# This is a 3D dataset

X_train.shape
```

```python
input_dimension = X_train.shape[1]*X_train.shape[2]
input_dimension
```

```python
# Flatten the training and testing data

X_train_flat = X_train.reshape(X_train.shape[0], -1)
X_test_flat = X_test.reshape(X_test.shape[0], -1)
```

```python
# Scale the dataset

X_train_flat_scaled = X_train_flat / 255
X_test_flat_scaled = X_test_flat / 255
```

```python
# Design the model

num_classes = len(np.unique(y_test))

model = Sequential()

model.add(Dense(128, activation='relu', input_dim=input_dimension))
model.add(Dense(num_classes, activation='softmax'))
```

```python
model.compile(loss='sparse_categorical_crossentropy', optimizer='Adam')
```

```python
# Fit based on reshaped X

model.fit(X_train_flat_scaled, y_train, epochs=10, validation_split=0.2)
```

```python
# Predict based on reshaped X
sample_probs = model.predict(X_test_flat_scaled)

sample_probs[:5]
```

```python
# See the shape of the sample_probs

sample_probs[0].shape
```

```python
# Find the classification for each sample

y_pred = np.argmax(sample_probs, axis=1)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score

print(f"Accuracy score: {accuracy_score(y_test, y_pred)}")
```
